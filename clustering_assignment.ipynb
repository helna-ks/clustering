Importing libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

Load Iris Dataset

```python
iris = load_iris()
iris
```

Coverting to dataframe

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
df.head()
```

Adding the species column

```python
df['species'] = iris.target
df.head()
```

```python
X=df.drop('species', axis=1)
```

```python
X.head()
```

Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

  Question No 1 =  K-Means Clustering

**Working of K-means Clustering** - K-Means divides the data into a
fixed number of clusters. It assigns each data point to the nearest cluster center and repeatedly updates the centers until the clusters become stable.

**K-Means Suitable for Iris** - K-Means is suitable because the Iris dataset has numerical features that can be grouped based on similarity. We can use 3 clusters because Iris contains three types of flowers.

```python
from sklearn.cluster import KMeans
```

creating K-Means model

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
```

```python
kmeans.fit(X_scaled)
```

```python
kmeans_labels = kmeans.labels_
print(kmeans_labels)
```

Visualize clusters

```python
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=kmeans_labels)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('K-Means Clustering')
plt.show()
```

Question No - 2 = Hierarchical Clustering

**Working of Hierarchical Clustering** - Hierarchical clustering creates groups by gradually joining similar data points together. It forms a tree-like structure called a dendrogram.

Hierarchical Clustering suitable for Iris - It is suitable because Iris has a small number of data points and different flower groups can be identified based on their similarities.

```python
from sklearn.cluster import AgglomerativeClustering
```

Creating the model

```python
hc = AgglomerativeClustering(n_clusters=3)
```

```python
hc_labels = hc.fit_predict(X_scaled)
print(hc_labels)
```

Visualize Hierarchical Clusters

```python
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=hc_labels)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Hierarchical Clustering')
plt.show()
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage
linked = linkage(X_scaled, method='ward')
plt.figure(figsize=(10, 5))
dendrogram(linked)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Data Points')
plt.ylabel('Distance')
plt.show()
```


**K-Means**: Groups the Iris data into 3 clusters using distance from cluster centers.

**Hierarchical**: Groups similar Iris data points step by step and represents them using a dendrogram.

Both methods are suitable because the Iris dataset contains numerical features with clear groups
